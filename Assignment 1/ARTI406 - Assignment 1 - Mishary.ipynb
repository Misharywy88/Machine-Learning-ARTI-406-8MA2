#### **Univariate Analysis** 📊  

##### *Ramen Ratings Distribution* 🍜

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Stars'], bins=20)
plt.title("Distribution of Ramen Ratings")
plt.xlabel("Stars")
plt.ylabel("Count")
plt.show()
```

##### *Top 10 Countries by Number of Reviews* 🌍🏆

```python
plt.figure(figsize=(10,5))
df['Country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Countries by Number of Reviews")
plt.ylabel("Count")
plt.show()
```

#### **Bivariate Analysis** 🔗

##### *Top 10 Countries - Average Rating* ✨

```python
plt.figure(figsize=(10,5))

df.groupby('Country')['Stars'].mean().sort_values(ascending=False).head(10).plot(kind='bar')

plt.title("Top 10 Countries by Average Rating")
plt.ylabel("Average Stars")
plt.show()
```

##### *Japan vs Korea vs USA* 

```python
# 7. Japan vs Korea vs USA 
df[df['Country'].isin(['Japan','South Korea','United States'])]\
  .groupby('Country')['Stars'].mean().plot(kind='bar')
plt.show()
```
